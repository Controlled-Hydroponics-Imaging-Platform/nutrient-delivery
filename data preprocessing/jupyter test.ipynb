```python
import csv
import os

# Tank geometry (rectangular, ref CAD on Fusion360)
tank_height_mm = 275            # tank height in mm
tank_area_m2 = 0.17646          # cross-sectional area in m²

# Input and output file paths
input_path = os.path.join(os.getcwd(),"water_level_data.csv") 
output_path = os.path.join(os.getcwd(), "water_volume_data.csv")

# Open input CSV and process each row
with open(input_path, mode='r', newline='') as infile, \
     open(output_path, mode='w', newline='') as outfile:
    
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    # Read and write header
    header = next(reader)
    timestamp_index = header.index("timestamp")
    distance_index = header.index("distance_mm")
    writer.writerow(["timestamp", "volume_L"])  # new header

    # Process each row
    for row in reader:
        try:
            timestamp = row[timestamp_index]
            distance_mm = float(row[distance_index])
            
            # Compute water height from bottom (in mm), then convert to meters
            water_height_mm = tank_height_mm - distance_mm
            water_height_m = water_height_mm / 1000.0

            # Calculate volume (in L)
            volume_L = water_height_m * tank_area_m2 * 1000.0

            # Write result
            writer.writerow([timestamp, round(volume_L, 3)])

        except (ValueError, IndexError):
            # Skip rows with invalid data
            continue

print("Processed data saved to:", output_path)
# end
```
